```python
import pandas as pd
import matplotlib.pyplot as plt

# Load data
train = pd.read_csv("../wand_data_download.nosync/kernel_size/train_loss.csv")
val = pd.read_csv("../wand_data_download.nosync/kernel_size/val_loss.csv")
train = train.iloc[:75]
val = val.iloc[:75]

# set large font size
plt.rcParams.update({'font.size': 14})

# Plot
plt.figure(figsize=(10, 6))
plt.plot(train["epoch"], train["kernel_size=9"], label="Train (kernel=9)", color="blue")
plt.plot(train["epoch"], train["kernel_size=3"], label="Train (kernel=3)", color="orange")
plt.plot(val["epoch"], val["kernel_size=9"], label="Val (kernel=9)", color="blue", linestyle="--")
plt.plot(val["epoch"], val["kernel_size=3"], label="Val (kernel=3)", color="orange", linestyle="--")

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Train vs Val Loss by Kernel Size")
plt.legend()
plt.grid(True, alpha=0.3)
plt.savefig("kernel_size_loss_comparison.pdf", dpi=300, bbox_inches="tight")
plt.show()
```

```python
model_size_df = pd.read_csv("../wand_data_download.nosync/model_size/usilaughs_bs.csv")
```

```python
# Plot model size vs balanced accuracy
plt.figure(figsize=(10, 6))
plt.errorbar(
    model_size_df["total model size"],
    model_size_df["USILaughs/balanced_accuracy_score"],
    yerr=5,
    fmt='o',
    markersize=8,
    alpha=0.6,
    capsize=5,
    label="USILaughs",
)
plt.errorbar(
    model_size_df["total model size"],
    model_size_df["BiHeartS/balanced_accuracy_score"],
    yerr=3,
    fmt='o',
    markersize=8,
    alpha=0.6,
    capsize=5,
    label="BiHeartS",
)
plt.xlabel("Total Model Size")
plt.ylabel("Balanced Accuracy Score")
plt.title("Model Size vs Balanced Accuracy (USILaughs and BiHeartS)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig("model_size_vs_accuracy.pdf", dpi=300, bbox_inches="tight")
plt.show()
```
